### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
def level(x):
  if x >= 90:
    return 'High'
  elif x>=70:
    return 'medium'
  else:
    return 'Low'
```

```python
df['scoring_level'] = df['total_score'].apply(level)
```
